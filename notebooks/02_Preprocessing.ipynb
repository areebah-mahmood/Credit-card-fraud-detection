Load and remove duplicates

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

df = pd.read_csv("/content/drive/MyDrive/creditcard.csv")
df = df.drop_duplicates()
print(df.shape)
```

Split into features and label

```python
X = df.drop("Class", axis=1)
y = df["Class"]
```

Train/Test split

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)

print(X_train.shape, X_test.shape)
print(y_train.mean(), y_test.mean())
```

Scale Time and Amount

```python
scaler = StandardScaler()
cols = ["Time", "Amount"]
X_train[cols] = scaler.fit_transform(X_train[cols])
X_test[cols] = scaler.transform(X_test[cols])
```

Save the results

```python
X_train.to_csv("/content/drive/MyDrive/X_train.csv", index=False)
X_test.to_csv("/content/drive/MyDrive/X_test.csv", index=False)
y_train.to_csv("/content/drive/MyDrive/y_train.csv", index=False)
y_test.to_csv("/content/drive/MyDrive/y_test.csv", index=False)
```

## Summary

- **Data cleaning:** removed 1,081 duplicate rows, leaving 283,726 rows and 31 columns.
- **Features and target:** `X` holds the 30 input features and `y` holds `Class` (0 = normal, 1 = fraud).
- **Train/test split:** 80% train (226,980 rows) and 20% test (56,746 rows), stratified so the fraud rate (~0.17%) is the same in both sets.
- **Feature scaling:** `Time` and `Amount` standardized with `StandardScaler`, fit on the training set only to avoid data leakage. V1–V28 were already scaled.
- **Output:** saved `X_train`, `X_test`, `y_train` and `y_test` as CSV files in Google Drive for use in the next notebooks.
